## Atividade 1
Escreva os métodos de ponto fixo dentro do arquivo $utils/algoritmos.py$.

````python
import numpy as np
from scipy.optimize import approx_fprime

def pontofixo(a, g, TOL=1e-8):
    x = g(a)
    while abs(x - a) > TOL:
        a = x
        x = g(a)
    return x

def newton_raphson(a, f, TOL=1e-8, df=None):
    if df is None:
        def dfn(x):
            return approx_fprime(np.array([x]), lambda v: f(v[0]))[0]
    else:
        dfn = df
    g = lambda x: x - f(x) / dfn(x)
    return pontofixo(a, g, TOL)

def secante(a, b, f, TOL=1e-8):
    g = lambda a, b: (a * f(b) - b * f(a)) / (f(b) - f(a))
    x = g(a, b)
    while abs(x - b) > TOL:
        a, b = b, x
        x = g(a, b)
    return x
````

Resolver a equação $e^x = x + 2$ é equivalente a calcular os pontos fixos da função  

$$
g(x) = e^x - 2
$$

Use os métodos do ponto fixo  

$$
x^{(n+1)} = g(x^{(n)})
$$

com $x^{(0)} = -1.8$ para obter uma aproximação de uma das soluções da equação dada com 8 dígitos significativos.  

**Resposta:**  
$
x \approx -1.8414057
$

````python
import numpy as np

# Define a função de iteração g(x)
def g(x):
    return np.exp(x) - 2

# Define a função de ponto fixo fornecida
def pontofixo(a, g, TOL=1e-8):
    x = g(a)
    while abs(x - a) > TOL:
        a = x
        x = g(a)
    return x

# Define os parâmetros iniciais
x0 = -1.8
tolerancia = 1e-8 # Para garantir precisão de 8 casas decimais

# Executa o método
raiz_aproximada = pontofixo(x0, g, TOL=tolerancia)

# Imprime o resultado formatado com 8 casas decimais
print(f"Aproximação da solução: x ≈ {raiz_aproximada:.8f}")
````

## Atividade 2
Encontre a raiz positiva da função  

$$
f(x) = \cos(x) - x^2
$$  

pelos métodos do ponto fixo, inicializando-o com $x^{(0)} = 1$.  

Realize a iteração até obter estabilidade no **quinto dígito significativo**.  

**Resposta:**

$$
x \approx 0.82413 
$$

Processo iterativo:  

$$
x^{(n+1)} = x^{(n)} + \frac{\cos(x) - x^2}{\sin(x) + 2x}
$$

````python
import numpy as np

# Define a função de iteração g(x) conforme o processo iterativo da imagem
def g(x):
    return x + (np.cos(x) - x**2) / (np.sin(x) + 2*x)

# Define a função do método do ponto fixo (da sua biblioteca)
def pontofixo(a, g, TOL=1e-8):
    x = g(a)
    while abs(x - a) > TOL:
        a = x
        x = g(a)
    return x

# Define os parâmetros iniciais
x0 = 1.0
tolerancia = 1e-5 # Para garantir a estabilidade no quinto dígito significativo

# Executa o método
raiz_aproximada = pontofixo(x0, g, TOL=tolerancia)

# Imprime o resultado formatado com 5 casas decimais
print(f"Aproximação da raiz: x ≈ {raiz_aproximada:.5f}")
````

## Atividade 3
Aplique os métodos do ponto fixo para resolver a equação:

$$
e^{-x^2} = 2x
$$

**Resposta:**

$$
x \approx 0.4193648
$$

````python
import numpy as np

# Define a função de iteração g(x) isolando o x
def g(x):
    return np.exp(-x**2) / 2

# Define a função do método do ponto fixo (da sua biblioteca)
def pontofixo(a, g, TOL=1e-8):
    x = g(a)
    while abs(x - a) > TOL:
        a = x
        x = g(a)
    return x

# Define os parâmetros iniciais
x0 = 0.5 # Valor inicial arbitrário próximo da raiz
tolerancia = 1e-8 # Para garantir a precisão mostrada na resposta

# Executa o método
raiz_aproximada = pontofixo(x0, g, TOL=tolerancia)

# Imprime o resultado formatado com 7 casas decimais
print(f"Aproximação da raiz: x ≈ {raiz_aproximada:.7f}")
````
